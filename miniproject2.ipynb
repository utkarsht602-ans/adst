```python
import pandas as pd
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
data = {
    'age': [25,45,31,52,22],
}
df = pd.DataFrame(data)
print(df)
```

```python
scaler = MinMaxScaler()
scaled = scaler.fit_transform(df[['age']])
df_scaled = pd.DataFrame(scaled, columns=['age_scaled'])
print(df_scaled)
```
